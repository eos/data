```python
import eos
import numpy as np
import logging
eos.logger.setLevel(logging.ERROR)
```

```python
def weighted_average(values, weights, axis=0):
    average = np.average(values, weights=weights, axis=axis)
    variance = np.average((values-average)**2, weights=weights, axis=axis)
    return np.array((average, np.sqrt(variance)))
```

```python
af = eos.AnalysisFile("analysis_BSM.yaml")
```

```python
for p in af.posteriors:
    data = eos.ImportanceSamples(f"data/{p}/samples")
    CVL_samples = data.samples[:,-2]
    CVL_central, CVL_err = weighted_average(CVL_samples, data.weights)
    print(f"{p}: C_VL = {CVL_central:.4f} \\pm {CVL_err:#.2g}")
```

```python
for p in af.posteriors:
    data = eos.ImportanceSamples(f"data/{p}/samples")
    CVR_samples = data.samples[:,-1]
    CVR_central, CVR_err = weighted_average(CVR_samples, data.weights)
    print(f"{p}: C_VR = {CVR_central:.4f} \\pm {CVR_err:#.2g}")
```

```python
for p in af.posteriors:
    data = eos.ImportanceSamples(f"data/{p}/samples")
    samples = data.samples[:,(-2,-1)]
    weights = data.weights
    central = np.average(samples, weights=weights, axis=0)
    cov = np.cov(samples.T, aweights=weights)
    unc = np.sqrt(np.diag(cov))
    corr = cov / np.outer(unc, unc)
    print(p)
    print(central, unc)
    print(corr)
```
